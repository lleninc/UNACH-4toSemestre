```python
#==========================================================
#  ALGORITMO ACO (Colonia de Hormigas) - Ejemplo Simple
#  ¿Cuál es el camino más corto al azúcar?
# ============================================================

import random

# ------------------------------------------------------------
# 1. EL ESCENARIO (Dos opciones de ruta)
# ------------------------------------------------------------
caminos = ["Camino A (Muy largo)","Camino B (Promedio)","Camino C (Super corto)","Camino D (Rapido)",]
distancias = [20, 10, 2, 8]
feromonas = [1.0, 1.0, 1.0, 1.0]  # Empezamos con rastro igual (caos)

# ------------------------------------------------------------
# 2. PARÁMETROS
# ------------------------------------------------------------
N_HORMIGAS  = 30          # Hormigas por iteración
ITERACIONES = 15           # Cuántas veces saldrá la colonia
EVAPORACION = 0.35        # Evaporación moderada (rho)
Q           = 80          # Cantidad de "pintura" de la hormiga

# ------------------------------------------------------------
# 3. CICLO DE LA COLONIA
# ------------------------------------------------------------
print("--- 🐜 INICIO DE LA COLONIA ---")

for i in range(1, ITERACIONES + 1):
    elecciones = []

    # PASO 1: Las hormigas eligen camino basado en probabilidad
    for h in range(N_HORMIGAS):
        # Probabilidad = feromona / suma de todas las feromonas
        total_f = sum(feromonas)
        probabilidades = [f / total_f for f in feromonas]

        # Sorteo: La hormiga elige un camino (0 o 1)
        eleccion = random.choices(range(len(caminos)), weights=probabilidades)[0]
        elecciones.append(eleccion)

    # PASO 2: Evaporación (El sol borra el rastro)
    for j in range(len(feromonas)):
        feromonas[j] *= (1 - EVAPORACION)

    # PASO 3: Depósito (Refuerzo por las hormigas que pasaron)
    for eleccion in elecciones:
        # Aquí está la clave: Q / L (mucha pintura si el camino es corto)
        refuerzo = Q / distancias[eleccion]
        feromonas[eleccion] += refuerzo

    # Reporte de la iteración
    votos_a = elecciones.count(0)
    votos_b = elecciones.count(1)
    votos_c = elecciones.count(2)
    votos_d = elecciones.count(3)
    print(f"Iteración {i}: A={votos_a} hormigas | B={votos_b} hormigas | C={votos_c} hormigas | D={votos_d} hormigas | Feromonas: {feromonas}")
    
# ------------------------------------------------------------
# 4. RESULTADO FINAL
# ------------------------------------------------------------
print("\n--- 🏁 CONCLUSIÓN ---")
ganador = 0 if feromonas[0] > feromonas[1] else 1
ganador = 0 if feromonas[0] > feromonas[1] else (1 if feromonas[1] > feromonas[2] else (2 if feromonas[2] > feromonas[3] else 3))
print(f"La colonia decidió que la mejor ruta es: {caminos[ganador]}")
```
