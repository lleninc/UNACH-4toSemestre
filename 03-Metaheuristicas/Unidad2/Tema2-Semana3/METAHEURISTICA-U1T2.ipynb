### Algoritmo PSO (Particle Swarm Optimization)

```python
import random
import time

# Función objetivo
def funcion(x, y):
    return 3*x**2 + 2*y**2

# Parámetros
num_particulas = 10
iteraciones = 50
limite_inferior = -10
limite_superior = 10

# Inicialización
particulas = []
velocidades = []

for i in range(num_particulas):
    x = random.uniform(limite_inferior, limite_superior)
    y = random.uniform(limite_inferior, limite_superior)
    particulas.append([x, y])
    velocidades.append([0, 0])

# Mejor global
mejor_global = particulas[0]
mejor_valor = funcion(mejor_global[0], mejor_global[1])

inicio = time.time()

# Algoritmo PSO
for _ in range(iteraciones):
    for i in range(num_particulas):
        x, y = particulas[i]

        valor = funcion(x, y)

        if valor < mejor_valor:
            mejor_valor = valor
            mejor_global = [x, y]

        # Actualizar velocidad (simplificado)
        velocidades[i][0] = random.random() * (mejor_global[0] - x)
        velocidades[i][1] = random.random() * (mejor_global[1] - y)

        # Actualizar posición
        particulas[i][0] += velocidades[i][0]
        particulas[i][1] += velocidades[i][1]

fin = time.time()

# Resultados
print("PSO:")
print("Mejor solución:", mejor_global)
print("Valor mínimo:", mejor_valor)
print("Tiempo:", fin - inicio)
```

### Algoritmo Genético (GA)

```python
import random
import time

# Función objetivo
def funcion(x, y):
    return 3*x**2 + 2*y**2

# Parámetros
tam_poblacion = 10
generaciones = 50
limite_inferior = -10
limite_superior = 10

# Crear población inicial
poblacion = []
for _ in range(tam_poblacion):
    x = random.uniform(limite_inferior, limite_superior)
    y = random.uniform(limite_inferior, limite_superior)
    poblacion.append([x, y])

inicio = time.time()

# Algoritmo Genético
for _ in range(generaciones):
    # Evaluar población
    poblacion.sort(key=lambda ind: funcion(ind[0], ind[1]))

    # Selección (mejores 50%)
    mejores = poblacion[:tam_poblacion//2]

    # Cruce
    nuevos = []
    while len(nuevos) < tam_poblacion:
        padre1 = random.choice(mejores)
        padre2 = random.choice(mejores)

        hijo_x = (padre1[0] + padre2[0]) / 2
        hijo_y = (padre1[1] + padre2[1]) / 2

        # Mutación
        if random.random() < 0.1:
            hijo_x += random.uniform(-1, 1)
            hijo_y += random.uniform(-1, 1)

        nuevos.append([hijo_x, hijo_y])

    poblacion = nuevos

fin = time.time()

# Mejor solución
mejor = min(poblacion, key=lambda ind: funcion(ind[0], ind[1]))

print("GA:")
print("Mejor solución:", mejor)
print("Valor mínimo:", funcion(mejor[0], mejor[1]))
print("Tiempo:", fin - inicio)
```

### Algoritmo ACO (Ant Colony Optimization)

```python
import random
import time

# Función objetivo
def funcion(x, y):
    return 3*x**2 + 2*y**2

# Parámetros
num_hormigas = 10
iteraciones = 50
limite_inferior = -10
limite_superior = 10

# Inicialización
mejor_solucion = None
mejor_valor = float('inf')

inicio = time.time()

# Algoritmo ACO (simplificado)
for _ in range(iteraciones):
    soluciones = []

    for _ in range(num_hormigas):
        x = random.uniform(limite_inferior, limite_superior)
        y = random.uniform(limite_inferior, limite_superior)

        valor = funcion(x, y)
        soluciones.append((x, y, valor))

        if valor < mejor_valor:
            mejor_valor = valor
            mejor_solucion = (x, y)

fin = time.time()

print("ACO:")
print("Mejor solución:", mejor_solucion)
print("Valor mínimo:", mejor_valor)
print("Tiempo:", fin - inicio)
```
