```python
import math
import random
import time

def calcular_costo(x):
    """Queremos minimizar f(x) = x^2. El mínimo ideal es x = 0."""
    return x ** 2

# =====================================================================
# SIMULATED ANNEALING CON CAPTURA DE MÉTRICAS
# =====================================================================
def algoritmo_recocido_simulado_metricas(temp_inicial, alfa, max_iteraciones):
    inicio_tiempo = time.perf_counter() # Inicia cronómetro
    
    solucion_actual = random.uniform(-10.0, 10.0)
    costo_actual = calcular_costo(solucion_actual)
    
    mejor_historico = solucion_actual
    costo_mejor_historico = costo_actual
    
    t = temp_inicial
    iteraciones_totales = 0
    iteraciones_sin_mejora = 0
    
    for i in range(1, max_iteraciones + 1):
        iteraciones_totales = i
        vecino = solucion_actual + random.uniform(-1.0, 1.0)
        vecino = max(-10.0, min(10.0, vecino))
        costo_vecino = calcular_costo(vecino)
        
        delta = costo_vecino - costo_actual
        
        if delta < 0:
            solucion_actual = vecino
            costo_actual = costo_vecino
        else:
            probabilidad = math.exp(-delta / t)
            if random.uniform(0, 1) < probabilidad:
                solucion_actual = vecino
                costo_actual = costo_vecino
                
        # Verificar mejora real para la convergencia
        if costo_actual < costo_mejor_historico - 1e-6:
            mejor_historico = solucion_actual
            costo_mejor_historico = costo_actual
            iteraciones_sin_mejora = 0 # Reiniciar contador
        else:
            iteraciones_sin_mejora += 1
            
        # Criterio de convergencia: 20 iteraciones estancado
        if iteraciones_sin_mejora >= 20:
            break
            
        t = t * alfa
        
    fin_tiempo = time.perf_counter()
    tiempo_total = (fin_tiempo - inicio_tiempo) * 1000 # Convertir a milisegundos
    
    return mejor_historico, costo_mejor_historico, iteraciones_totales, tiempo_total

# =====================================================================
# TABU SEARCH CON CAPTURA DE MÉTRICAS
# =====================================================================
def algoritmo_busqueda_tabu_metricas(max_iteraciones, tamaño_lista_tabu):
    inicio_tiempo = time.perf_counter() # Inicia cronómetro
    
    solucion_actual = random.uniform(-10.0, 10.0)
    costo_actual = calcular_costo(solucion_actual)
    
    mejor_historico = solucion_actual
    costo_mejor_historico = costo_actual
    
    lista_tabu = []
    iteraciones_totales = 0
    iteraciones_sin_mejora = 0
    
    for i in range(1, max_iteraciones + 1):
        iteraciones_totales = i
        candidatos = [solucion_actual + random.uniform(-1.0, 1.0) for _ in range(5)]
        candidatos = [max(-10.0, min(10.0, c)) for c in candidatos]
        
        candidatos_permitidos = [c for c in candidatos if round(c, 1) not in lista_tabu]
        
        if not candidatos_permitidos:
            candidatos_permitidos = candidatos
            
        mejor_candidato = min(candidatos_permitidos, key=calcular_costo)
        costo_candidato = calcular_costo(mejor_candidato)
        
        solucion_actual = mejor_candidato
        costo_actual = costo_candidato
        
        lista_tabu.append(round(solucion_actual, 1))
        if len(lista_tabu) > tamaño_lista_tabu:
            lista_tabu.pop(0)
            
        # Verificar mejora real para la convergencia
        if costo_actual < costo_mejor_historico - 1e-6:
            mejor_historico = solucion_actual
            costo_mejor_historico = costo_actual
            iteraciones_sin_mejora = 0
        else:
            iteraciones_sin_mejora += 1
            
        # Criterio de convergencia: 20 iteraciones estancado
        if iteraciones_sin_mejora >= 20:
            break
            
    fin_tiempo = time.perf_counter()
    tiempo_total = (fin_tiempo - inicio_tiempo) * 1000 # Convertir a milisegundos
    
    return mejor_historico, costo_mejor_historico, iteraciones_totales, tiempo_total

# =====================================================================
# BLOQUE DE EJECUCIÓN PRINCIPAL
# =====================================================================
if __name__ == "__main__":
    # Fijamos la semilla para que los resultados de tu reporte sean estables
    random.seed(15) 
    
    # Ejecutar Simulated Annealing
    sol_sa, costo_sa, iter_sa, tiempo_sa = algoritmo_recocido_simulado_metricas(
        temp_inicial=100.0, alfa=0.95, max_iteraciones=500
    )
    
    # Ejecutar Tabu Search
    sol_ts, costo_ts, iter_ts, tiempo_ts = algoritmo_busqueda_tabu_metricas(
        max_iteraciones=500, tamaño_lista_tabu=5
    )
    
    print("=== MÉTRICAS GENERADAS CON ÉXITO ===")
    print(f"\n[SA] Iteraciones: {iter_sa} | Solución X: {sol_sa:.6f} | Tiempo: {tiempo_sa:.4f} ms")
    print(f"[TS] Iteraciones: {iter_ts} | Solución X: {sol_ts:.6f} | Tiempo: {tiempo_ts:.4f} ms")
```
