## 📈 Regresión lineal vs polinómica

```python
import numpy as np # librería para manejo de arrays y generación de datos
import matplotlib.pyplot as plt # librería para visualización de datos
from sklearn.linear_model import LinearRegression # modelo de regresión lineal
from sklearn.preprocessing import PolynomialFeatures # para generar características polinómicas
from sklearn.metrics import mean_squared_error # para evaluar el modelo
 
# Generar datos np.random.seed(0) 
X = np.linspace(0, 10, 30).reshape(-1, 1) 
# Agregar ruido a los datos para simular una relación no lineal
# np.sin(X).ravel() genera una relación no lineal entre X e y, 
# y np.random.normal agrega ruido para hacer el problema más realista.

y = np.sin(X).ravel() + np.random.normal(scale=0.3, size=X.shape[0]) 

# Regresión lineal 
# Ajustamos un modelo de regresión lineal a los datos.
lin_reg = LinearRegression().fit(X, y) 
# Predecimos los valores de y utilizando el modelo lineal ajustado.
y_pred_lin = lin_reg.predict(X) 
 
# Regresión polinómica 
# Para la regresión polinómica, primero transformamos las características de X 
# para incluir términos polinómicos hasta el grado 5.
poly5 = PolynomialFeatures(degree=5) 
X_poly5 = poly5.fit_transform(X) # Transformamos X para incluir términos polinómicos hasta el grado 5.
poly5_reg = LinearRegression().fit(X_poly5, y) # Ajustamos un modelo de regresión lineal a las características polinómicas.
y_pred_poly5 = poly5_reg.predict(X_poly5) # Predecimos los valores de y utilizando el modelo polinómico ajustado.  
 
poly10 = PolynomialFeatures(degree=20) 
X_poly10 = poly10.fit_transform(X) # Transformamos X para incluir términos polinómicos hasta el grado 10.
poly10_reg = LinearRegression().fit(X_poly10, y) # Ajustamos un modelo de regresión lineal a las características polinómicas.
y_pred_poly10 = poly10_reg.predict(X_poly10) # Predecimos los valores de y utilizando el modelo polinómico ajustado.  
```

```python
# Gráfica comparativa
# Visualizamos los datos originales y las predicciones de ambos modelos para comparar su ajuste a los datos.
# Creamos una figura con dos subgráficos para comparar la regresión lineal con la polinómica de grado 5 y grado 20.

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5)) 

# Gráfico 1: Regresión Lineal vs Polinómica grado 5
ax1.scatter(X, y, color="black", label="Datos", s=30) # puntos de datos originales
ax1.plot(X, y_pred_lin, label="Lineal", color="blue", linewidth=2)# línea de regresión lineal
ax1.plot(X, y_pred_poly5, label="Polinómica grado 5", color="red", linewidth=2)# línea de regresión polinómica
ax1.set_xlabel("X") # etiqueta del eje x
ax1.set_ylabel("y") # etiqueta del eje y
ax1.set_title("Regresión Lineal vs Polinómica (grado 5)")
ax1.legend()
ax1.grid(True, alpha=0.3)#

# Gráfico 2: Regresión Lineal vs Polinómica grado 20
ax2.scatter(X, y, color="black", label="Datos", s=30)
ax2.plot(X, y_pred_lin, label="Lineal", color="blue", linewidth=2)
ax2.plot(X, y_pred_poly10, label="Polinómica grado 20", color="green", linewidth=2)
ax2.set_xlabel("X")
ax2.set_ylabel("y")
ax2.set_title("Regresión Lineal vs Polinómica (grado 20)")
ax2.legend()
ax2.grid(True, alpha=0.3)

plt.tight_layout() # ajusta el espacio entre los subgráficos para que no se solapen
plt.show()
```

```python
# Calculamos el error cuadrático medio (MSE) para ambos modelos para cuantificar su rendimiento.
from sklearn.metrics import mean_squared_error 
 
mse_lin = mean_squared_error(y, y_pred_lin) # Calculamos el MSE para el modelo de regresión lineal.
mse_poly5 = mean_squared_error(y, y_pred_poly5) # Calculamos el MSE para el modelo de regresión polinómica de grado 5.
print("MSE Lineal:", mse_lin) # Imprimimos el MSE del modelo de regresión lineal.
print("MSE Polinómica:", mse_poly5) # Imprimimos el MSE del modelo de regresión polinómica de grado 5.

```

### Riesgo al aumentar el grado del polinomio

<div style="border-left: 6px solid #f59e0b; background:#fffbeb; padding:12px 14px; border-radius:8px; margin:10px 0;">
  <strong style="color:#b45309;">ADVERTENCIA</strong><br>
  Al aumentar demasiado el grado del polinomio aparece el <strong>sobreajuste</strong>.
</div>

<p>El modelo aprende muy bien los datos de entrenamiento, pero también captura el ruido y pierde capacidad de generalización.</p>

<p><strong>Señales típicas del sobreajuste:</strong></p>
<ul>
  <li>Error de entrenamiento muy bajo.</li>
  <li>Error de validación o prueba más alto.</li>
  <li>Curva demasiado ondulada para seguir casi todos los puntos.</li>
</ul>

### Relación con el dilema sesgo-varianza

<div style="border-left: 6px solid #0ea5e9; background:#f0f9ff; padding:12px 14px; border-radius:8px; margin:10px 0;">
  <strong style="color:#0369a1;">IMPORTANTE</strong><br>
  El dilema sesgo-varianza explica por qué no siempre conviene usar un polinomio de grado alto.
</div>

<ul>
  <li><strong>Grado bajo (modelo simple):</strong> alto sesgo y baja varianza. Puede subajustar (underfitting).</li>
  <li><strong>Grado alto (modelo complejo):</strong> bajo sesgo y alta varianza. Puede sobreajustar (overfitting).</li>
</ul>

<p><strong>Conclusión:</strong> el objetivo es encontrar un grado intermedio que minimice el error en datos no vistos, equilibrando sesgo y varianza.</p>

##  📉 Regularización (Ridge y Lasso) 

```python
# Ejemplo de regularización con Ridge y Lasso
from sklearn.linear_model import Ridge, Lasso # modelos de regresión con regularización
from sklearn.datasets import make_regression # para generar datos de regresión sintéticos

# Generar datos sintéticos para regresión 
X, y = make_regression(n_samples=100, n_features=10, noise=10, random_state=0) #
ridge = Ridge(alpha=1.0).fit(X, y) # Ajustamos un modelo de regresión Ridge a los datos con un valor de alpha de 1.0.
lasso = Lasso(alpha=0.1).fit(X, y) # Ajustamos un modelo de regresión Lasso a los datos con un valor de alpha de 0.1.
 
print("Coeficientes Ridge:", ridge.coef_) 
print("Coeficientes Lasso:", lasso.coef_) 
```

###  ¿Por qué Lasso puede considerarse un método de selección de variables?

<div style="border-left: 6px solid #f59e0b; background:#fffbeb; padding:12px 14px; border-radius:8px; margin:10px 0;">
  <strong style="color:#b45309;">POR QUE?</strong><br>
  Lasso se considera un método de selección de variables por su capacidad de realizar escasez <strong>(sparsity) </strong>.
</div>

<p>A diferencia de Ridge, que solo reduce el tamaño de los coeficientes, la penalización de Lasso (basada en el valor absoluto, norma L1) puede forzar a que los coeficientes de las variables menos importantes lleguen a exactamente cero.</p>

<p><strong>Datos importantes:</strong></p>
<ul>
  <li>Cuando un coeficiente es cero, esa variable queda efectivamente "fuera" del modelo.</li>
  <li>Simplifica el modelo automáticamente sin necesidad de que un humano decida qué columnas borrar.</li>
  <li>Selecciona solo el subconjunto de predictores que realmente impactan en la respuesta.</li>
</ul>

## Regresión logística y Softmax 

```python
from sklearn.datasets import load_iris 
from sklearn.linear_model import LogisticRegression 
from sklearn.metrics import classification_report 
 
iris = load_iris() 
X, y = iris.data, iris.target 
log_reg = LogisticRegression(max_iter=200).fit(X, y) 
y_pred = log_reg.predict(X) 
print(classification_report(y, y_pred, target_names=iris.target_names)) 
```

###  SVM y Redes Neuronales 

```python
# Ejemplo de clasificación con SVM y MLP
from sklearn.datasets import make_moons # para generar datos de clasificación sintéticos con forma de luna
from sklearn.svm import SVC # para el modelo de máquina de vectores de soporte (SVM)
from sklearn.neural_network import MLPClassifier # para el modelo de perceptrón multicapa (MLP) para clasificación
 
X, y = make_moons(n_samples=200, noise=0.2, random_state=0) #Conjunto de datos sintéticos con forma de luna para clasificación.
# Ajustamos un modelo de máquina de vectores de soporte (SVM) con un kernel radial (RBF) a los datos.
svm = SVC(kernel="rbf", C=1, gamma=0.5).fit(X, y) # svm gamma=0.5 para un ajuste más flexible a los datos no lineales.
mlp = MLPClassifier(hidden_layer_sizes=(5,), max_iter=1000, # max_iter=1000 para asegurar la convergencia del modelo de MLP.
random_state=0).fit(X, y) # Ajustamos un modelo de perceptrón multicapa (MLP) a los datos.
print("Accuracy SVM:", svm.score(X, y)) 
print("Accuracy MLP:", mlp.score(X, y)) 
```

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_moons
from sklearn.svm import SVC
from sklearn.neural_network import MLPClassifier

# 1. Generación de datos
X, y = make_moons(n_samples=200, noise=0.2, random_state=0)

# 2. Entrenamiento de modelos
svm = SVC(kernel="rbf", C=1, gamma=0.5).fit(X, y)
mlp = MLPClassifier(hidden_layer_sizes=(5,), max_iter=1000, random_state=0).fit(X, y)

# 3. Creación de la malla para dibujar las fronteras
h = .02  # tamaño del paso en la malla
x_min, x_max = X[:, 0].min() - 0.5, X[:, 0].max() + 0.5
y_min, y_max = X[:, 1].min() - 0.5, X[:, 1].max() + 0.5
xx, yy = np.meshgrid(np.arange(x_min, x_max, h), np.arange(y_min, y_max, h))

# 4. Configuración de la figura
names = ["SVM (Kernel RBF)", "MLP (Perceptrón Multicapa)"]
models = [svm, mlp]

plt.figure(figsize=(20, 5))

for i, model in enumerate(models):
    plt.subplot(1, 2, i + 1)
    
    # Predecir sobre la malla
    Z = model.predict(np.c_[xx.ravel(), yy.ravel()]) # Predecimos las clases para cada punto en la malla.
    Z = Z.reshape(xx.shape)

    # Dibujar la frontera de decisión (fondo de color)
    plt.contourf(xx, yy, Z, cmap=plt.cm.RdBu, alpha=0.8)
    
    # Dibujar los puntos de los datos reales
    plt.scatter(X[:, 0], X[:, 1], c=y, edgecolors='k', cmap=plt.cm.RdBu)
    
    plt.title(f"{names[i]}\nAccuracy: {model.score(X, y):.2f}")
    plt.xlabel("Característica 1")
    plt.ylabel("Característica 2")

plt.tight_layout()
plt.show()
```

### ¿Cuál modelo resulta más flexible y por qué? 

<div style="border-left: 6px solid #aff50b; background:#fffbeb; padding:12px 14px; border-radius:8px; margin:10px 0;">
  <strong style="color: #09b41a;"> ✅ FLEXIBLE</strong><br>
  <strong style="color: #094b0f;">  El SVM con kernel RBF tiende a ser más flexible.</strong><br>
 
</div>

<p>El kernel RBF (Radial Basis Function) permite al SVM crear fronteras de decisión curvas y suaves.</p>

<p><strong>el MLP está limitado por tener: </strong></p>
<ul>
  <li>Solo una capa oculta con 5 neuronas (hidden_layer_sizes=(5,))</li>
  <li>onstruye la frontera mediante la intersección de planos (líneas).</li>
  <li>El modelo no tiene ninguna dificultad en trazar una "línea" que la aísle perfectamente de las demás.</li>
</ul>

### ¿Qué ventaja tiene la SVM frente a la red neuronal? 

<div style="border-left: 6px solid #e9b60e; background: #f0f9ff; padding:12px 14px; border-radius:8px; margin:10px 0;">
  <strong style="color: #a19603;"> VENTAJAS</strong><br>
  <strong style="color: #181616;"> Eficiencia en datasets pequeños: Con solo 200 muestras, la SVM es extremadamente robusta</strong><br>
  
</div>

<ul>
  <li><strong>Encuentra el "margen máximo" de separación, lo que garantiza una mejor generalización con pocos datos.</strong> </li>
  <li><strong>Menos propensa a mínimos locales</strong> La optimización de la SVM es un problema matemático convexo; esto significa que siempre encontrará la solución óptima global.</li>
</ul>

<p><strong>Conclusión:</strong>Es más fácil entender el impacto de C y gamma en una SVM que ajustar la arquitectura (capas y neuronas) y el aprendizaje en un MLP.</p>

  <strong style="color: #03a108;"> DONDE USAR MLP</strong><br>
